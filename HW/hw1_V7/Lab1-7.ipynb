````python
import pandas as pd
````

````python
df = pd.read_csv('data.csv')
````

1. Какой средний вес людей представленных в наборе данных?

````python
print(f"Средний вес: {df['Weight (kg)'].mean()} кг")
````

2. Сбалансировано ли соотношение мужчин и женщин в наборе данных?

````python
gender_ratio = df['Gender'].value_counts(normalize=True) * 100
print(gender_ratio)
is_balanced = abs(gender_ratio.get('Male', 0) - 50) < 5
print(f"Сбалансировано: {'Да' if is_balanced else 'Нет'}")
````

3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
median_stress = df['Stress Levels'].median()
high_stress = df[df['Stress Levels'] > median_stress]
low_stress = df[df['Stress Levels'] <= median_stress]
bp_high = high_stress['Blood Pressure (s/d)'].str.split('/').str[0].astype(float).mean()
bp_low = low_stress['Blood Pressure (s/d)'].str.split('/').str[0].astype(float).mean()

print(f"с высоким стрессом {bp_high:.2f}")
print(f"с низким {bp_low:.2f}")
print(f"Правда ли: {'Да' if bp_high > bp_low else 'Нет'}")
````

4. Правда ли, что уровень холестерина у мужчин и женщин отличается?

````python
cholesterol_gender = df.groupby('Gender')['Cholesterol Level (mg/dL)'].mean()
print(cholesterol_gender)

diff = abs(cholesterol_gender.get('Male', 0) - cholesterol_gender.get('Female', 0))
print(f"Правда ли: {'Да' if diff > 1 else 'Нет'}")
````

5. Правда ли, что с возрастом женщины имеют свойство оценивать свои когнитивные способности хуже, чем в юности (порог юности - 30 лет, женщина в возрасте - 60 лет+)

````python
women = df[df['Gender'] == 'Female']
youth_cog = women[women['Age (years)'] < 30]['Cognitive Function'].mean()
elderly_cog = women[women['Age (years)'] >= 60]['Cognitive Function'].mean()
print(f"С<30: {youth_cog:.1f}")
print(f"60+: {elderly_cog:.1f}")
print(f"Правда ли: {'Да' if elderly_cog < youth_cog else 'Нет'}")
````

6. Влияет ли физическая активность на вес в возрасте более 50 лет?

````python
over_50 = df[df['Age (years)'] > 50]
weight_by_activity = over_50.groupby('Physical Activity Level')['Weight (kg)'].mean()
print(weight_by_activity)
````

7. Сравните минимальный вес мужчин с высоким и низким заработком

````python
m = df[df['Gender'] == 'Male']
a = m.loc[m['Income Level'] == 'Low', 'Weight (kg)'].min()
b = m.loc[m['Income Level'] == 'High', 'Weight (kg)'].min()
print('Low:', a)
print('High:', b)
````

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
p75 = df['BMI'].quantile(0.75)
p60 = df['BMI'].quantile(0.60)
p30 = df['BMI'].quantile(0.30)
df['Has_Disease'] = df['Chronic Diseases'].apply(lambda x: 0 if pd.isna(x) or str(x).lower() in ['none', 'no', '0'] else 1)
high_bmi_disease_rate = df[df['BMI'] > p75]['Has_Disease'].mean()
mid_bmi_disease_rate = df[(df['BMI'] >= p30) & (df['BMI'] <= p60)]['Has_Disease'].mean()

print(f"ИМТ >75: {high_bmi_disease_rate:.1%}")
print(f"ДИМТ 30-60: {mid_bmi_disease_rate:.1%}")
print(f"Правда ли: {'Да' if high_bmi_disease_rate > mid_bmi_disease_rate else 'Нет'}")
````

9. Правда ли, что люди с высоким уровнем дохода имеют больший уровень физической активности, чем люди с низким доходом

````python
df_filtered = df[df['Income Level'].isin(['High', 'Low'])].copy()
activity_mapping = {'Low': 1, 'Moderate': 2, 'High': 3}
df_filtered['Activity_Score'] = df_filtered['Physical Activity Level'].map(activity_mapping)

mean_activity = df_filtered.groupby('Income Level')['Activity_Score'].mean()
print(f"низкий:  {mean_activity['Low']}")
print(f"высокий: {mean_activity['High']}")
print(f"Правда ли: {'Да' if mean_activity['High'] > mean_activity['Low'] else 'Нет'}")
````

10. Какая доля мужчин с высокой частотой пребывания на солнце имеет высокий заработок?

````python
m = df[df['Gender'] == 'Male']
limit = df['Sun Exposure'].quantile(.75)
selected = m[m['Sun Exposure'] > limit]
print('доля:', round(selected['Income Level'].eq('High').mean() * 100, 2), '%')
````

5. Постройте круговую диаграмму отображающую как ментальное здоровье связано с паттернами сна

````python
counts = df.groupby(['Mental Health Status', 'Sleep Patterns']).size().unstack(fill_value=0)
mental_health_categories = ['Poor', 'Fair', 'Good', 'Excellent']
sleep_categories = ['Normal', 'Insomnia', 'Excessive']
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()
for idx, category in enumerate(mental_health_categories):
    data = counts.loc[category, sleep_categories]
    axes[idx].pie(
        data.values,
        labels=data.index,
        autopct='%1.1f%%',
        startangle=90
    )
    axes[idx].set_title(f'Ментальное здоровье: {category}', fontweight='bold')
plt.suptitle("Распределение паттернов сна по уровню ментального здоровья", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

6. Имеет ли связь зрение и способность слышать звуки на высоких частотах? Ответьте на вопрос - какая зависимость между ними и предположите почему зависимость именно такая?

````python
plt.figure(figsize=(8, 5))
plt.scatter(df['Vision Sharpness'], df['Hearing Ability (dB)'], alpha=0.3)
plt.title('Связь зрения и слуха')
plt.xlabel('Четкость зрения')
plt.ylabel('Слух (дБ)')
plt.show()
corr = df['Vision Sharpness'].corr(df['Hearing Ability (dB)'])
print(f"Коэффициент корреляции: {corr:.3f}")
````

На графике видна обратная зависимость - чем лучше зрение, тем лучше слух. Возможно из-за старения оба показателя одновременно падают
